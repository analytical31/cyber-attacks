Visual Analysis

```python
# 1. Failed Attacks by Country
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("security_logs.csv")

failed = df[df["status"] == "Failed"]

plt.figure(figsize=(28,4))

plt.subplot(2,4,1)
failed["country"].value_counts().plot(kind="bar", color="red")
plt.title("Failed Attacks by Country")
```

```python
# 2. Login Status - success vs failed


plt.subplot(2,4,2)

plt.figure(figsize=(28,4))
df["status"].value_counts().plot(kind="pie", autopct="%1.1f%%")

plt.title("Login status - Success vs Failed")
```

```python
# 3. Top Users Failed Attacks

plt.subplot(2,4,3)

plt.figure(figsize=(10,4))

failed["user_name"].value_counts().head(5).plot(kind="bar", color='green')
plt.title("Top Users Failed Attacks")
```

```python
# 4. Attacks by Country

plt.subplot(2,4,4)

plt.figure(figsize=(10,4))

failed["country"].value_counts().plot(kind="line", marker="o")
plt.title("Attacks by Country ")
```

```python
# 5. On Risk Countries

plt.subplot(2,4,5)

plt.figure(figsize=(10,4))

failed["country"].value_counts().plot(kind="bar", color='orange')
plt.title("On Risk Countries")
```

```python
# 6. Users - facing attacks

plt.subplot(2,4,6)

plt.figure(figsize=(10,4))

failed["user_name"].value_counts().plot(kind="pie", autopct="%1.1f%%")
plt.title("User Attacks")
```

```python
# 7. Top IP"S by Attack

plt.subplot(2,4,7)

plt.figure(figsize=(10,4))

failed["ip_address"].value_counts().head(10).plot(kind="bar", color='purple')
plt.title("Top IPs")
```

```python
# 8. Failed login status

plt.subplot(2,4,8)

plt.figure(figsize=(10,4))

failed["status"].value_counts().plot(kind="bar", color="orange")
plt.title("Failed Status")

plt.tight_layout()

plt.show()
```

```python
# 9. Login trend
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("security_logs.csv")

df["login_time"] = pd.to_datetime(
    df["login_time"]
)

monthly = df.groupby(
    df["login_time"].dt.month
).size()

monthly.plot(kind="line")

plt.title("Monthly Login Trend")
plt.xlabel("Month")
plt.ylabel("Number of Logins")

plt.show()
```
